```python
import matplotlib.pyplot as plt
cause_counts = df["Cause"].value_counts()
plt.figure(figsize=(8,5))
plt.bar(cause_counts.index,cause_counts.values)
plt.xlabel("cause")
plt.ylabel("Number of accidents")
plt.xticks(rotation=30)
plt.show
```
